## scipy で ODE を解く

簡単な指数減衰の ODE を解いて、結果をプロットする。

$$\frac{dy}{dt} = -0.5 y, \quad y(0) = 1$$

解析解: $y(t) = e^{-0.5t}$

```python
from scipy.integrate import solve_ivp


def decay(t, y):
    return -0.5 * y


# 数値解
sol = solve_ivp(decay, t_span=[0, 10], y0=[1.0], dense_output=True)

# プロット
t = np.linspace(0, 10, 200)
y_numerical = sol.sol(t)[0]
y_analytical = np.exp(-0.5 * t)

plt.figure(figsize=(8, 5))
plt.plot(t, y_numerical, label="solve_ivp", linewidth=2)
plt.plot(t, y_analytical, "--", label="analytical: e^(-0.5t)", linewidth=2)
plt.xlabel("t")
plt.ylabel("y")
plt.title("Exponential decay: dy/dt = -0.5 y")
plt.legend()
plt.grid(True)
plt.show()
```
